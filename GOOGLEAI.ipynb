
###  6 : COMBINING


```python

#Load and prepare main weather DataFrame
df = pd.read_csv('weatherHistory.csv')
df['Formatted Date'] = pd.to_datetime(df['Formatted Date'], utc=True)
df['Month'] = df['Formatted Date'].dt.month

# Aggregate monthly averages from the main dataset
monthly_weather = df.groupby('Month').agg({
    'Temperature (C)': 'mean',
    'Humidity': 'mean',
    'Wind Speed (km/h)': 'mean'
}).reset_index()

#  Create a second DataFrame containing monthly weather categories
season_data = {
    'Month': list(range(1, 13)),
    'Season': [
        'Winter', 'Winter', 'Spring', 'Spring', 'Spring', 'Summer',
        'Summer', 'Summer', 'Autumn', 'Autumn', 'Autumn', 'Winter'
    ],
    'Category': [
        'Cold/Dry', 'Cold/Dry', 'Mild', 'Mild/Rainy', 'Warm', 'Hot',
        'Hot', 'Hot/Dry', 'Warm', 'Cool/Rainy', 'Cool', 'Cold'
    ]
}
category_df = pd.DataFrame(season_data)

# Merge the two DataFrames on the 'Month' column
merged_df = pd.merge(monthly_weather, category_df, on='Month', how='inner')

# Display result
print("--- Merged Monthly Weather DataFrame ---")
print(merged_df)
```
